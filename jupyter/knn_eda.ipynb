# KNN Empirical P-value EDA

Analysis of KNN-based empirical P-values for window comparison results.

**Comparison:** {comparison}

**Data:** results/window_compare/{comparison}/{comparison}_knn_enriched.tsv


```python
import os, sys, importlib.util

# Load knn_viz by file path — avoids "jupyter" built-in package conflict
_knn_viz_path = os.path.join(os.getcwd(), "knn_viz.py")
_spec = importlib.util.spec_from_file_location("_knn_viz", _knn_viz_path)
_knn_viz = importlib.util.module_from_spec(_spec)
_spec.loader.exec_module(_knn_viz)
sys.modules["_knn_viz"] = _knn_viz

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from _knn_viz import KNNVizPlotter, auto_discover, load_data

os.makedirs("results/figures", exist_ok=True)
print("Available comparisons:", auto_discover("/share/home/xhhuang24/project/PanTE/SNAP-HiC-Diff/results/window_compare"))

```

```python
# Change COMPARISON here to switch analysis
COMPARISON = "H1398_vs_L134"
df = load_data(COMPARISON)
print(f"Loaded {len(df):,} windows for {COMPARISON}")
print("Chromosomes:", sorted(df["chrom"].unique()))
```

```python
plotter = KNNVizPlotter(df, COMPARISON, "results/figures")
```

## 1. KDE per Score


```python
plotter.kde_per_score()
plt.show()
```

## 2. Chrom KDE Heatmap


```python
for col in KNNVizPlotter.SCORE_COLS:
    plotter.chrom_kde_heatmap(col)
plt.show()
```

## 3. Pairwise Hexbin Density


```python
plotter.hexbin_pairs()
plt.show()
```

## 4. KNN Distance vs P-value


```python
plotter.knn_vs_score()
plt.show()
```

## 5. PCA with Density Contour


```python
plotter.pca_contour()
plt.show()
```

## 6. P-value Correlation Heatmap


```python
plotter.corr_heatmap()
plt.show()
```

## 7. Background Deviation KDE


```python
plotter.bg_dev_kde()
plt.show()
```

## 8. Summary Statistics


```python
plotter.summary_table()
plt.show()
```
